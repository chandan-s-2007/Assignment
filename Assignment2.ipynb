```python
#4.Write a function that takes a number and returns whether it is even or odd.
def ev():
    a=int(input("Enter a number:"))
    if a%2==0:
        print("Even")
    else:
        print("Odd")
ev()
```

```python
#5.Write a function that takes three numbers and returns the largest without using max().
def lar():
    a=int(input("Enter number1:"))
    b=int(input("Enter number2:"))
    c=int(input("Enter number3:"))
    if a>b and b>c:
        print("number1 is largest")
    elif c>b:
        print("number3 is largest")
    else:
        print("number2 is largest")
lar() 
```

```python
#7.Write a function that takes a year and returns whether it is a leap year.
def leap():
    a=int(input("Enter year:"))
    if a%4==0:
        print("Leap year")
    else:
        print("not a leap year")
leap()
```

```python
#9.Write a function that takes three numbers and returns the smallest without using min().
def sm():
    a=int(input("Enter number1:"))
    b=int(input("Enter number2:"))
    c=int(input("Enter number3:"))
    if a<b and b<c:
        print("number1 is smallest")
    elif c<b:
        print("number3 is smallest")
    else:
        print("number2 is smallest")
sm() 
```

```python
#10.Write a function that takes a number and returns how many digits it contains.
def dig():
    a=input("Enter a number:")
    sum=0
    for i in a:
        sum+=1
    print("Number of digits:",sum)
dig()
```

```python
#11.Write a function that takes a number and returns the product of its digits.
def pd():
    a=input("Enter a number:")
    sum=1
    for i in a:
        sum*=int(i)
    print("Product of digits:",sum)
pd()
```

```python
#12.Write a function that checks whether a number is divisible by both 3 and 5.
def div():
    a=int(input("Enter a number:"))
    if a%3==0 and a%5==0:
        print("Divisible by both 3 and 5")
    elif a%5==0:
        print("Divisible only by 5")
    elif a%3==0:
        print("Divisible only by 3")
    else:
        print("Not divisible by 3 or 5")
div()
```
