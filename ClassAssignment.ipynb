```python
#From [-12, 5, -7, 18, 0, 21, -3, 10], select numbers divisible by 3, convert negative numbers to positive, square them, and calculate their total.

lis=[-12, 5, -7, 18, 0, 21, -3, 10]
from functools import reduce
def div(x):
    if x%3==0:
        return x
def neg(x):
    if x<0:
        return -x
    return x
def sq(x):
    return x*x
def add(x,y):
    return x+y
r=list(filter(div,lis))
r1=list(map(neg,r))
r2=list(map(sq,r1))
r3=reduce(add,r2)
print(r3)
```

```python
#From [14, 25, 30, 42, 55, 60, 73, 90], find the largest number that is divisible by 5 but not divisible by 3.
lis=[14, 25, 30, 42, 55, 60, 73, 90]
def div(x):
    if x%5==0 and x%3!=0:
        return x
def lar(x,y):
    if x>y:
        return x
    else:
        return y
r=list(filter(div,lis))
r1=reduce(lar,r)
print(r1)
```

```python
#Use reduce() to find both the smallest and largest values from [34, 11, 78, 5, 46, 92, 17] in a single operation.
lis=[34, 11, 78, 5, 46, 92, 17]
from functools import reduce
def lar(x,y):
    if x>y:
        return x
    else:
        return y
def sm(x,y):
    if x<y:
        return x
    else:
        return y
r=reduce(lar,lis)
r1=reduce(sm,lis)
print("largest=",r)
print("smallest=",r1)
```

```python
#Calculate the difference between the largest and smallest numbers in [45, 12, 89, 23, 67, 4, 91] using only one reduce().
lis=[45, 12, 89, 23, 67, 4, 91] 
from functools import reduce
def lar(x,y):
    if x>y:
        return x
    else:
        return y
r=reduce(lar,lis)
print(r-4)
```

```python
#Calculate the average of [18, 25, 32, 47, 58] without using sum() or len().
lis=[18, 25, 32, 47, 58]
from functools import reduce
def avg(x,y):
    return x+y
r=reduce(avg,lis)
print(r/5)
```
