```python
#Take a username and password and check whether both match predefined values.
user="User1"
passw="1234"
u=input("Enter username:")
p=input("Enter password:")
if u==user and p==passw:
    print("Username and password matched")
elif u==user:
    print("Only username matched")
elif p==passw:
    print("Only password matched")
else:
    print("Username and password both doesn't match")
```
